```python
# 3) Load the model once
from clipnova_worker.playground import Playground, PRESETS
pg = Playground()
pg.load()
```

```python
#@title 4) Your prompt: edit the boxes, then press play
prompt = "A fluffy orange cat rides a skateboard down a sunny Indian street, handheld vlog style, bright morning light" #@param {type:"string"}
seconds = 2 #@param {type:"slider", min:1, max:5, step:1}
aspect = "9:16" #@param ["9:16", "16:9", "1:1", "4:5"]
steps = 20 #@param {type:"slider", min:8, max:50, step:2}
seed = 0 #@param {type:"integer"}
pg.play(prompt, seconds, aspect, steps, seed or None)
```

```python
# 5) (optional) Try all the starter prompts in one go
for p in PRESETS:
    print("\n>>", p)
    pg.play(p, seconds=2, aspect="9:16", steps=20)
```

```python
# 6) What to paste back to the team
print(pg.summary())
```

**Tips.** More steps = better but slower. Bad result? Change the seed or reword the prompt. A red message means the model could not make that clip; it says why.
